## Задание 11.3 — обработка данных о курсе акций


```python
stock_raw = pd.DataFrame({
    'Date': ['2026-09-21','2026-09-22','2026-09-23','2026-09-24','2026-09-25',
             '2026-09-28','2026-09-29','2026-09-30','2026-10-01','2026-10-02'],
    'Open':  ['$101.50', '$103.20', None, '$102.10', '$105.30', '$106.20', '$107.10', None, '$109.00', '$110.40'],
    'High':  ['$104.20', '$105.00', '$104.80', '$106.30', '$108.00', '$108.40', '$110.20', '$111.00', '$112.50', '$113.00'],
    'Low':   ['$100.80', '$101.70', '$100.50', '$101.20', None, '$104.80', '$105.90', '$107.00', '$108.20', '$109.50'],
    'Close': ['$103.80', '$102.40', '$103.10', '$105.90', '$107.20', None, '$109.40', '$110.10', '$111.70', '$112.30'],
    'Volume':[1200000, 980000, 1150000, None, 1300000, 1280000, 1400000, 1380000, None, 1500000]
})
stock_raw.to_csv('stocks_2026.csv', index=False)
display(stock_raw)
```

```python
# 1. Чтение CSV

def money_to_float(value):
    if pd.isna(value) or str(value).strip() == '':
        return np.nan
    text = str(value).replace('$', '').replace('€', '').replace('£', '').strip()
    return float(text)

stock = pd.read_csv(
    'stocks_2026.csv',
    header=0,
    usecols=['Date', 'Open', 'High', 'Low', 'Close', 'Volume'],
    parse_dates=['Date'],
    converters={
        'Open': money_to_float,
        'High': money_to_float,
        'Low': money_to_float,
        'Close': money_to_float
    },
    decimal='.'
)

stock['Volume'] = pd.to_numeric(stock['Volume'], errors='coerce').astype('Float64')

print(stock.dtypes)
display(stock)
```

```python
# 2а
sorted_stock = stock.sort_values('Volume', ascending=False)
display(sorted_stock)
```

```python
# 2б
stock['Close_rank'] = stock['Close'].rank(ascending=False, method='min')
display(stock[['Date', 'Close', 'Close_rank']])
```

```python
# 2в
display(stock[['Open', 'High', 'Low', 'Close']].describe())
```

```python
# 2г
pair = stock[['Close', 'Volume']].astype(float)

print('Корреляция:')
display(pair.corr())

print('Ковариация:')
display(pair.cov())
```

```python
# 2д
rows_to_drop = stock[stock.isna().sum(axis=1) >= 2]
print('Строки с 2 и более пропусками:')
display(rows_to_drop)

cleaned = stock.dropna(thresh=stock.shape[1] - 1).copy()

numeric_cols = ['Open', 'High', 'Low', 'Close', 'Volume']
for col in numeric_cols:
    cleaned[col] = cleaned[col].fillna(cleaned[col].mean())

print('После обработки пропусков:')
display(cleaned)
```

## Задание 11.3, пункт 3 — объединение данных за другой неперекрывающийся период


```python
stock_previous = pd.DataFrame({
    'Date': pd.to_datetime(['2026-09-16','2026-09-17','2026-09-18']),
    'Open': [98.0, 99.2, 100.1],
    'High': [100.0, 101.3, 102.2],
    'Low': [97.5, 98.7, 99.4],
    'Close': [99.4, 100.7, 101.8],
    'Volume': [900000, 950000, 1100000]
})

# 3а
concat_non_overlap = pd.concat(
    [stock_previous, stock.drop(columns=['Close_rank'])],
    ignore_index=True
)
print('concat() для неперекрывающихся периодов:')
display(concat_non_overlap)

# 3б
merge_non_overlap = pd.merge(
    stock_previous,
    stock.drop(columns=['Close_rank']),
    on=['Date', 'Open', 'High', 'Low', 'Close', 'Volume'],
    how='outer'
).sort_values('Date').reset_index(drop=True)

print('merge() для неперекрывающихся периодов:')
display(merge_non_overlap)
```

## Задание 11.3, пункт 4 — объединение данных за перекрывающийся период без дублирования


```python
stock_overlap = pd.DataFrame({
    'Date': pd.to_datetime(['2026-09-29','2026-09-30','2026-10-01','2026-10-05']),
    'Open': [107.1, 108.2, 109.0, 113.0],
    'High': [110.2, 111.0, 112.5, 115.0],
    'Low': [105.9, 107.0, 108.2, 112.0],
    'Close': [109.4, 110.1, 111.7, 114.2],
    'Volume': [1400000, 1380000, 1450000, 1600000]
})

current = stock.drop(columns=['Close_rank']).copy()

# 4а
concat_overlap = (
    pd.concat([current, stock_overlap], ignore_index=True)
      .drop_duplicates(subset='Date', keep='last')
      .sort_values('Date')
      .reset_index(drop=True)
)
print('concat() без дублирования:')
display(concat_overlap)

# 4б
merge_overlap = (
    current.set_index('Date')
    .combine_first(stock_overlap.set_index('Date'))
    .sort_index()
    .reset_index()
)
print('Реляционное объединение без дублирования:')
display(merge_overlap)
```

## Задание 11.3, пункт 5 — объединение с данными другой акции


```python
other_stock = pd.DataFrame({
    'Date': pd.to_datetime(['2026-09-21','2026-09-22','2026-09-23','2026-09-24','2026-09-25']),
    'Open': [55.0, 55.8, 56.1, 55.6, 57.0],
    'High': [56.2, 56.5, 57.0, 57.2, 58.1],
    'Low': [54.5, 55.1, 55.4, 55.0, 56.3],
    'Close': [55.9, 56.0, 56.7, 56.8, 57.6],
    'Volume': [700000, 720000, 690000, 750000, 810000]
})

stock_a = current.copy()
stock_a['Ticker'] = 'PETR'
stock_b = other_stock.copy()
stock_b['Ticker'] = 'OTHER'

# 5а.
two_stocks_concat = pd.concat([stock_a, stock_b], ignore_index=True)
print('Две акции через concat():')
display(two_stocks_concat)

# 5б.
two_stocks_merge = pd.merge(
    stock_a,
    stock_b,
    on='Date',
    how='outer',
    suffixes=('_PETR', '_OTHER')
).sort_values('Date').reset_index(drop=True)

print('Две акции через merge():')
display(two_stocks_merge)
```
